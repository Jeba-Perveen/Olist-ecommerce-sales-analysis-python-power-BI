# Importing libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# Importing the Tables

```python
customers = pd.read_csv('Olist Data/olist_customers_dataset.csv')
geolocation = pd.read_csv('Olist Data/olist_geolocation_dataset.csv')
order_items = pd.read_csv('Olist Data/olist_order_items_dataset.csv')
order_payments = pd.read_csv('Olist Data/olist_order_payments_dataset.csv')
order_reviews = pd.read_csv('Olist Data/olist_order_reviews_dataset.csv')
orders = pd.read_csv('Olist Data/olist_orders_dataset.csv')
products = pd.read_csv('Olist Data/olist_products_dataset.csv')
sellers = pd.read_csv('Olist Data/olist_sellers_dataset.csv')
category_name_translation = pd.read_csv('Olist Data/product_category_name_translation.csv')
```

```python
datasets = { "customer": customers,
            "geolocation":geolocation,
            "order_items":order_items,
            "order_payments":order_payments,
            "order_reviews":order_reviews,
            "orders":orders,
            "products":products,
            "sellers":sellers,
            "category_name_translation":category_name_translation}
for name,df in datasets.items():
    print(f"{name} : {df.shape}")
```

## Creating the base Table

```python
master_df = orders.copy()
```

```python
# merging customer table with orders table
```

```python
master_df = master_df.merge(customers, on = 'customer_id', how = 'left')
```

```python
master_df.shape
```

```python
master_df.isnull().sum()
```

```python
orders.shape
```

```python
customers.shape
```

```python
# merging master_df(customer and orders) with order_items table
master_df = master_df.merge(order_items,on = 'order_id',how='left')
```

```python
master_df.shape
```

```python
master_df['order_id'].nunique()
```

```python
# merging master table with products
master_df = master_df.merge(products,on='product_id',how='left')
```

```python
master_df.shape
```

```python
master_df.isnull().sum()
```

```python
# merging category translation table
master_df = master_df.merge(category_name_translation,on='product_category_name',how='left')
```

```python
# merging order_payments table
master_df = master_df.merge(order_payments,on='order_id',how='left')
```

```python
master_df.shape
```

```python
# merging reviews table
master_df = master_df.merge(order_reviews,on='order_id',how='left')
```

```python
master_df.shape
```

```python
# merging sellers table
master_df = master_df.merge(sellers,on='seller_id',how='left')
```

```python
master_df.shape
```

```python
# creating a copy of master_df 
master = master_df.copy()
```

```python
master.to_csv("master_table.csv",index=False)
```
